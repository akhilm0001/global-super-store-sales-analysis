```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
```

```python
#convert date of birth to datetime formst
df['date_of_birth'] = pd.to_datetime(df['date_of_birth'])
```

```python
df['date_of_birth'].dtype
```

```python
# date of birth to age
reference_date = pd.Timestamp('2026-01-01')

df['age'] = (reference_date.year - df['date_of_birth'].dt.year)
```

```python
df[df['age'] < 18][['customer_id', 'date_of_birth', 'age']]
```

```python
df['age'].describe()
```

```python
df[['date_of_birth', 'age']].head(10)
```

```python
labels = ['Under_18','Young_Adult', 'Adult', 'Middle_Aged', 'Senior']
```

```python
df['age_group'] = pd.cut(
    df['age'],
    bins=[0, 17, 25, 40, 60, 100],
    labels=labels
)
```

```python
df[['age','age_group']].head(20)
```

```python
df['age_group'].value_counts()
```

```python
df['age'].min(), df['age'].max()
```

```python
import matplotlib.pyplot as plt

age_group_counts = df['age_group'].value_counts()

plt.figure(figsize=(8,5))
plt.bar(age_group_counts.index, age_group_counts.values)
plt.title('Customer Distribution by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Number of Customers')
plt.xticks(rotation=45)
plt.show()
```

```python
# change datatype
df['order_date'] = pd.to_datetime(df['order_date'])
df['ship_date'] = pd.to_datetime(df['ship_date'])
df['sales_date'] = pd.to_datetime(df['sales_date'])
```

```python
df['year'] = df['order_date'].dt.year
df['quarter'] = df['order_date'].dt.quarter
df['month'] = df['order_date'].dt.month
df['month_name'] = df['order_date'].dt.month_name()
df['week'] = df['order_date'].dt.isocalendar().week
df['day'] = df['order_date'].dt.day
df['day_name'] = df['order_date'].dt.day_name()
```

```python
category_analysis = df.groupby('category_of_goods').agg(
    total_sales=('sales', 'sum'),
    total_profit=('profit', 'sum'),
    total_quantity=('quantity', 'sum')
).sort_values('total_sales', ascending=False)

category_analysis
```

```python
top_subcategories = (
    df.groupby('sub_category')['sales']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10,5))
plt.bar(top_subcategories.index, top_subcategories.values)
plt.title('Top 10 Sub-categories by Sales')
plt.xlabel('Sub-category')
plt.ylabel('Total Sales')
plt.xticks(rotation=45, ha='right')
plt.show()
```
